```python

# Python Basics: Complete Beginner Program

# 1. Variables and Data Types
name = input("Enter your name: ")
age = int(input("Enter your age: "))
number1 = float(input("Enter first number: "))
number2 = float(input("Enter second number: "))

print("\nHello,", name)
print("Age:", age)
print("First number:", number1)
print("Second number:", number2)

# 2. Arithmetic Operators
print("\n--- Arithmetic Operators ---")
print("Addition:", number1 + number2)
print("Subtraction:", number1 - number2)
print("Multiplication:", number1 * number2)

if number2 != 0:
    print("Division:", number1 / number2)
    print("Floor Division:", number1 // number2)
    print("Modulus:", number1 % number2)
else:
    print("Cannot divide by zero!")

print("Power:", number1 ** number2)

# 3. Comparison Operators
print("\n--- Comparison Operators ---")
print("Equal:", number1 == number2)
print("Not Equal:", number1 != number2)
print("Greater:", number1 > number2)
print("Less:", number1 < number2)

# 4. If-Else Conditions
print("\n--- If-Else ---")
if age >= 18:
    print("You are an adult.")
else:
    print("You are under 18.")

# 5. If-Elif-Else
print("\n--- Grade System ---")
marks = float(input("Enter your marks (0-100): "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 80:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 50:
    print("Grade D")
else:
    print("Fail")

# 6. Logical Operators
print("\n--- Logical Operators ---")
if age >= 18 and age <= 60:
    print("You are between 18 and 60.")
if age < 18 or age > 60:
    print("You are below 18 or above 60.")
if not age < 18:
    print("You are not under 18.")

# 7. Switch Alternative: Match-Case (Python 3.10+)
print("\n--- Match-Case ---")
choice = int(input("Enter 1 for Add, 2 for Subtract, 3 for Multiply: "))

match choice:
    case 1:
        print("Result:", number1 + number2)
    case 2:
        print("Result:", number1 - number2)
    case 3:
        print("Result:", number1 * number2)
    case _:
        print("Invalid choice")

# 8. For Loop
print("\n--- For Loop ---")
for i in range(1, 6):
    print(i)

# 9. While Loop
print("\n--- While Loop ---")
count = 1
while count <= 5:
    print(count)
    count += 1

# 10. Break and Continue
print("\n--- Break and Continue ---")
for i in range(1, 6):
    if i == 3:
        continue
    if i == 5:
        break
    print(i)

# 11. List (Similar to an Array)
print("\n--- List ---")
numbers = [10, 20, 30, 40, 50]

print("All numbers:", numbers)
print("First number:", numbers[0])
print("List length:", len(numbers))

numbers.append(60)
print("After adding:", numbers)

# 12. Function
print("\n--- Function ---")
def add(a, b):
    return a + b

result = add(number1, number2)
print("Sum using function:", result)

# 13. String Operations
print("\n--- Strings ---")
print("Uppercase:", name.upper())
print("Lowercase:", name.lower())
print("Name length:", len(name))

# 14. Even or Odd
print("\n--- Even or Odd ---")
if age % 2 == 0:
    print("Your age is even.")
else:
    print("Your age is odd.")

# 15. Nested If
print("\n--- Nested If ---")
if age >= 18:
    if age >= 60:
        print("You are a senior adult.")
    else:
        print("You are an adult under 60.")
else:
    print("You are a minor.")

print("\nProgram finished successfully!")
```
