Task 1  Modify the program to count the total number of nodes visited by IDDFS across all iterations
and compare it with the number of nodes visited by BFS on the same tree

```python

import networkx as nx
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from collections import deque

tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H', 'I'],
    'E': ['J'],
    'F': ['K'],
    'G': [],
    'H': [], 'I': [], 'J': [], 'K': []
}

# Depth Limited Search
def dls(node, goal, limit, visited_order):
    visited_order.append(node)

    if node == goal:
        return 'found'

    if limit == 0:
        return 'cutoff'

    cutoff_occurred = False

    for child in tree[node]:
        res = dls(child, goal, limit - 1, visited_order)

        if res == 'found':
            return 'found'

        if res == 'cutoff':
            cutoff_occurred = True

    return 'cutoff' if cutoff_occurred else 'failure'


# Iterative Deepening DFS
def iddfs(start, goal, max_depth=10):

    total_visited = 0

    for depth in range(max_depth + 1):
        order = []

        res = dls(start, goal, depth, order)

        count = len(order)
        total_visited += count

        print(f"Depth limit {depth}: visited {' '.join(order):<24} -> {res}")
        print("Nodes visited:", count)

        if res == 'found':
            print("Total IDDFS nodes visited:", total_visited)
            return depth, total_visited

        if res == 'failure':
            print("Total IDDFS nodes visited:", total_visited)
            return None, total_visited

    return None, total_visited


# Breadth First Search
def bfs(start, goal):

    queue = deque([start])
    visited_order = []

    while queue:
        node = queue.popleft()
        visited_order.append(node)

        if node == goal:
            break

        for child in tree[node]:
            queue.append(child)

    print("\nBFS visited order:", ' '.join(visited_order))
    print("Total BFS nodes visited:", len(visited_order))

    return len(visited_order)


# Run IDDFS
print("IDDFS, goal = J")

depth, iddfs_count = iddfs('A', 'J')

print("\nGoal found at depth:", depth)


# Run BFS
print("\nBFS, goal = J")

bfs_count = bfs('A', 'J')


# Compare results
print("\n----- COMPARISON -----")
print("IDDFS total nodes visited:", iddfs_count)
print("BFS total nodes visited:", bfs_count)

if iddfs_count > bfs_count:
    print("BFS visited fewer nodes.")
elif bfs_count > iddfs_count:
    print("IDDFS visited fewer nodes.")
else:
    print("Both visited the same number of nodes.")


# Visualize tree
G = nx.DiGraph(
    [(p, c) for p, cs in tree.items() for c in cs]
)

pos, next_x = {}, [0]

def layout(n, d=0):
    kids = tree[n]

    if not kids:
        pos[n] = (next_x[0], -d)
        next_x[0] += 1
    else:
        for k in kids:
            layout(k, d + 1)

        pos[n] = (
            sum(pos[k][0] for k in kids) / len(kids),
            -d
        )

layout('A')

NAVY = '#0E367B'

colors = [
    'white' if n != 'J' else NAVY
    for n in G.nodes
]

plt.figure(figsize=(6, 3.6))

nx.draw(
    G, pos,
    with_labels=False,
    node_color=colors,
    node_size=700,
    edgecolors=NAVY,
    edge_color=NAVY,
    arrows=False
)

nx.draw_networkx_labels(
    G, pos,
    font_weight='bold',
    font_color={
        n: 'white' if n == 'J' else NAVY
        for n in G.nodes
    }
)

plt.title("Search Tree (Goal J Highlighted)")
plt.savefig('lab04_tree.png', dpi=130, bbox_inches='tight')

print("\nTree saved to lab04_tree.png")
```

Task 2  Visualise the search pattern: for each depth limit, save an image in which visited nodes are
coloured green and unvisited nodes grey (use nx.draw with a colour list).

```python

import networkx as nx
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H', 'I'],
    'E': ['J'],
    'F': ['K'],
    'G': [],
    'H': [], 'I': [], 'J': [], 'K': []
}

# Create graph
G = nx.DiGraph(
    [(p, c) for p, cs in tree.items() for c in cs]
)

# Create tree layout
pos, next_x = {}, [0]

def layout(n, d=0):
    kids = tree[n]

    if not kids:
        pos[n] = (next_x[0], -d)
        next_x[0] += 1
    else:
        for k in kids:
            layout(k, d + 1)

        pos[n] = (
            sum(pos[k][0] for k in kids) / len(kids),
            -d
        )

layout('A')


# Visualization function
def draw_tree(visited, depth):

    colors = [
        'green' if n in visited else 'lightgrey'
        for n in G.nodes
    ]

    plt.figure(figsize=(7, 4))

    nx.draw(
        G, pos,
        with_labels=True,
        node_color=colors,
        node_size=800,
        font_weight='bold',
        edgecolors='black',
        arrows=False
    )

    plt.title(f"IDDFS Search - Depth Limit {depth}")

    filename = f"iddfs_depth_{depth}.png"

    plt.savefig(filename, dpi=130, bbox_inches='tight')
    plt.close()

    print("Image saved:", filename)


# Depth Limited Search
def dls(node, goal, limit, visited_order):

    visited_order.append(node)

    if node == goal:
        return 'found'

    if limit == 0:
        return 'cutoff'

    cutoff_occurred = False

    for child in tree[node]:

        res = dls(child, goal, limit - 1, visited_order)

        if res == 'found':
            return 'found'

        if res == 'cutoff':
            cutoff_occurred = True

    return 'cutoff' if cutoff_occurred else 'failure'


# Iterative Deepening DFS
def iddfs(start, goal, max_depth=10):

    total_visited = 0

    for depth in range(max_depth + 1):

        order = []

        res = dls(start, goal, depth, order)

        total_visited += len(order)

        print(f"\nDepth limit {depth}")
        print("Visited nodes:", ' '.join(order))
        print("Result:", res)

        # Save image for this depth limit
        draw_tree(order, depth)

        if res == 'found':
            print("\nGoal found at depth:", depth)
            print("Total nodes visited:", total_visited)
            return

        if res == 'failure':
            return


# Run program
iddfs('A', 'J')
```

Task 3 Apply IDDFS to the 8-puzzle from Lab 2 for the start state (1, 2, 3, 4, 0, 6, 7, 5, 8). Print the
depth at which the goal is found and the number of nodes expanded.

```python

import time

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

START = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

MOVES = {
    'Up': -3,
    'Down': 3,
    'Left': -1,
    'Right': 1
}


# Generate possible actions
def actions(state):
    i = state.index(0)
    possible = []

    if i >= 3:
        possible.append('Up')

    if i < 6:
        possible.append('Down')

    if i % 3 != 0:
        possible.append('Left')

    if i % 3 != 2:
        possible.append('Right')

    return possible


# Apply a move
def result(state, move):
    i = state.index(0)
    j = i + MOVES[move]

    new_state = list(state)
    new_state[i], new_state[j] = new_state[j], new_state[i]

    return tuple(new_state)


# Depth Limited Search
def dls(state, depth, path, expanded):

    if state == GOAL:
        return path

    if depth == 0:
        return None

    expanded[0] += 1

    for move in actions(state):

        new_state = result(state, move)

        if new_state not in path:
            answer = dls(
                new_state,
                depth - 1,
                path + [new_state],
                expanded
            )

            if answer is not None:
                return answer

    return None


# Iterative Deepening DFS
def iddfs(start, max_depth=30):

    total_expanded = 0
    start_time = time.time()

    for depth in range(max_depth + 1):

        expanded = [0]

        answer = dls(
            start,
            depth,
            [start],
            expanded
        )

        total_expanded += expanded[0]

        print("Depth limit:", depth)
        print("Nodes expanded:", expanded[0])

        if answer is not None:

            end_time = time.time()

            print("\nGoal found!")
            print("Solution depth:", depth)
            print("Total nodes expanded:", total_expanded)
            print("Time taken:", round(end_time - start_time, 6), "seconds")

            return answer

    print("Goal not found within maximum depth.")
    print("Total nodes expanded:", total_expanded)


# Run IDDFS
iddfs(START)
```
